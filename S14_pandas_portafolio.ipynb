## 00 · ¿Qué es pandas y para qué se usa?

> **Nota** — Objetivo: entender qué problema resuelve pandas, y por qué no basta con lo que ya sabes de NumPy.

Un portafolio real no es solo números: tiene tickers (texto), cantidades (enteros), precios (decimales) y, más adelante, fechas. NumPy asume que todo el array es del mismo tipo — por eso necesitas algo que junte columnas de distinto tipo en una sola tabla, con nombres. Eso es exactamente lo que resuelve **pandas**.

pandas es una librería para trabajar con datos organizados en tablas: leerlos, explorarlos, limpiarlos, filtrarlos y resumirlos. Está construida sobre NumPy — por dentro, cada columna de una tabla de pandas es un array.

> 💡 **Idea clave** — NumPy responde *"¿cómo calculo rápido sobre muchos números?"*. pandas responde *"¿cómo organizo datos de distinto tipo en algo que se pueda leer, filtrar y consultar como una tabla?"*. No compiten — pandas usa NumPy por dentro.

📚 Fuente: ["Package overview", pandas.pydata.org](https://pandas.pydata.org/docs/getting_started/overview.html)

## 01 · La estructura: Series y DataFrame

> **Nota** — Objetivo: distinguir las dos estructuras de datos que ofrece pandas.

pandas ofrece dos tipos de estructuras:

- **Series** — un arreglo unidimensional etiquetado, capaz de contener datos de cualquier tipo. Es una sola columna, con índice.
- **DataFrame** — una estructura de dos dimensiones que guarda datos como una tabla, con filas y columnas. Cada columna de un DataFrame es, de hecho, una Series.

````python
import numpy as np
import pandas as pd

precios = pd.Series([195, 385, 185, 165])
print(precios)
print(type(precios))
````

````python
portafolio = pd.DataFrame({
    "ticker": ["AAPL", "MSFT", "AMZN", "GOOGL"],
    "cantidad": [10, 8, 12, 6],
    "precio": [195, 385, 185, 165]
})
print(portafolio)
print(type(portafolio))
````

## 06 · Todo junto: el portafolio como DataFrame

> **Nota** — Objetivo: integrar creación, columnas calculadas, filtro y orden en un solo flujo.

````python
portafolio = pd.DataFrame({
    "ticker": ["AAPL", "MSFT", "AMZN", "GOOGL"],
    "cantidad": [10, 8, 12, 6],
    "precio_compra": [180, 400, 170, 150],
    "precio_actual": [195, 385, 185, 165]
})

portafolio["valor_actual"] = portafolio["cantidad"] * portafolio["precio_actual"]
portafolio["ganancia"] = portafolio["valor_actual"] - (portafolio["cantidad"] * portafolio["precio_compra"])
portafolio["rendimiento"] = (portafolio["ganancia"] / (portafolio["cantidad"] * portafolio["precio_compra"])) * 100

en_perdida = portafolio[portafolio["rendimiento"] < 0]
portafolio.sort_values(by="rendimiento", ascending=False)
````
