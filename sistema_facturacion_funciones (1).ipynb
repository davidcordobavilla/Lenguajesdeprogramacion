Un diccionario para los productos y otro para los clientes.

````python
def ver_catalogo():
    """Muestra todos los productos del catálogo."""
    print("CATÁLOGO DE PRODUCTOS")
    # .items() entrega el código y los datos de cada producto.
    for codigo, producto in catalogo.items():
        print(codigo, "-", producto["nombre"], "· $", producto["precio"])

# Para verla funcionando:
ver_catalogo()
````

## 🎁 Cómo se lee una función

```
def  nombre_de_la_funcion(parametros):   ← qué recibe
     """Qué hace"""                     ← documentación
     ...cuerpo...                        ← qué hace
     return resultado                    ← qué devuelve (opcional)
```
Una función que solo **muestra** información no necesita `return`.

## 📖 Misión 1 · `ver_catalogo()`

Convierte el diccionario del catálogo en una función que lo muestre.
No recibe nada, no devuelve nada: solo imprime.

````python
def agregar_producto(codigo, nombre, precio):
    """Agrega un producto nuevo al catálogo."""
    # Un diccionario nuevo se guarda usando el código como llave.
    catalogo[codigo] = {"nombre": nombre, "precio": precio, "iva": 0.19}
    print("Producto agregado:", nombre)

# Prueba:
agregar_producto("T-040", "Memoria USB 32 GB", 38000)
ver_catalogo()
````

## ➕ Misión 2 · `agregar_producto()`

Recibe código, nombre y precio, y guarda el producto en el catálogo.
Recuerda: `catalogo[codigo] = {...}` agrega una llave nueva al diccionario.

````python
def agregar_producto(codigo, nombre, precio):
    """Agrega un producto nuevo al catálogo."""
    # Un diccionario nuevo se guarda usando el código como llave.
    catalogo[codigo] = {"nombre": nombre, "precio": precio, "iva": 0.19}
    print("Producto agregado:", nombre)

# Prueba:
agregar_producto("T-040", "Memoria USB 32 GB", 38000)
ver_catalogo()
````

## 👥 Misión 3 · `ver_clientes()`

La misma idea de la misión 1, pero con el diccionario de clientes.

````python
def ver_clientes():
    """Muestra todos los clientes registrados."""
    print("CATÁLOGO DE CLIENTES")
    for documento, cliente in clientes.items():
        print(documento, "-", cliente["nombre"], "·", cliente["tipo"])

ver_clientes()
````

## 🗑️ Misión 4 · `eliminar_cliente()`

Recibe el documento y borra ese cliente con `del clientes[documento]`.
Antes de borrar, revisa con un `if` que exista, para que el programa no se caiga.

````python
def eliminar_cliente(documento):
    """Elimina un cliente si existe."""
    if documento in clientes:
        del clientes[documento]          # del borra la llave del diccionario
        print("Cliente eliminado:", documento)
    else:
        print("Ese cliente no existe.")

eliminar_cliente("900456")
ver_clientes()
````

## 🔍 Misión 5 · `consultar()`

Recibe un dato y lo busca primero en productos y luego en clientes.
Si no está en ninguno, avisa que no lo encontró.

````python
def consultar(dato):
    """Busca el dato en el catálogo de productos o en el de clientes."""
    if dato in catalogo:
        print("Producto:", catalogo[dato]["nombre"], "· $", catalogo[dato]["precio"])
    elif dato in clientes:
        print("Cliente:", clientes[dato]["nombre"])
    else:
        print("No encontré", dato)

consultar("T-040")
consultar("900123")
````

## 🖥️ Misión 6 · `menu()` — la integradora

Aquí se juntan todas tus funciones. El menú ya no hace el trabajo: solo **dibuja las opciones**, pregunta y **llama** a la función correspondiente.

Así se verá en pantalla:
```
================================
   EL LÁPIZ FELIZ · CAJA
================================
   1. Ver catálogo
   2. Agregar producto
   3. Ver clientes
   4. Consultar
   5. Salir
```


````python
def menu():
    """Menú de la caja: solo coordina, no calcula."""
    opcion = ""
    while opcion != "5":
        print("=" * 32)
        print("   EL LÁPIZ FELIZ · CAJA")
        print("=" * 32)
        print("   1. Ver catálogo")
        print("   2. Agregar producto")
        print("   3. Ver clientes")
        print("   4. Consultar")
        print("   5. Salir")
        opcion = input("   Opción: ")

        if opcion == "1":
            ver_catalogo()
        elif opcion == "2":
            codigo = input("Código: ")
            nombre = input("Nombre: ")
            precio = float(input("Precio: "))
            agregar_producto(codigo, nombre, precio)
        elif opcion == "3":
            ver_clientes()
        elif opcion == "4":
            consultar(input("Código o documento: "))
        elif opcion == "5":
            print("   Hasta luego 👋")
        else:
            print("   Opción no válida")

menu()
````
